```python
import pandas as pd

df = pd.read_csv("../Dataset/fraudTest.csv")

print("Dataset loaded successfully!")
```

```python
print("Shape:", df.shape)
display(df.head())
```

```python
print("Total Columns:", len(df.columns))

for col in df.columns:
    print(col)
```

```python
df.info()
```

```python
missing = df.isnull().sum()

print(missing[missing > 0])
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
# Fraud distribution

fraud_counts = df['is_fraud'].value_counts()

print("Fraud Distribution:")
print(fraud_counts)

print("\nFraud Percentage:")
print(df['is_fraud'].value_counts(normalize=True) * 100)
```

```python
# Visualize fraud distribution

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='is_fraud')

plt.title('Fraud vs Non-Fraud Transactions')
plt.xlabel('Fraud (0 = No, 1 = Yes)')
plt.ylabel('Number of Transactions')

plt.show()
```

```python
print("Unique Cardholders:", df['cc_num'].nunique())

print("Unique Merchants:", df['merchant'].nunique())

print("Unique Categories:", df['category'].nunique())

print("Unique Cities:", df['city'].nunique())
```

```python
# Convert transaction timestamp to datetime

df['trans_date_trans_time'] = pd.to_datetime(
    df['trans_date_trans_time']
)

print(df['trans_date_trans_time'].dtype)
```

```python
df['hour'] = df['trans_date_trans_time'].dt.hour

df['day'] = df['trans_date_trans_time'].dt.day

df['month'] = df['trans_date_trans_time'].dt.month

df['day_name'] = df['trans_date_trans_time'].dt.day_name()

df[['trans_date_trans_time', 'hour', 'day_name', 'month']].head()
```

```python
fraud_by_hour = df.groupby('hour')['is_fraud'].sum()

plt.figure(figsize=(12,5))

sns.barplot(x=fraud_by_hour.index, y=fraud_by_hour.values)

plt.title('Fraud Transactions by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Fraud Transactions')

plt.show()
```

```python
hourly_analysis = df.groupby('hour').agg(
    total_transactions=('is_fraud', 'count'),
    fraud_transactions=('is_fraud', 'sum')
)

hourly_analysis['fraud_rate'] = (
    hourly_analysis['fraud_transactions'] /
    hourly_analysis['total_transactions']
) * 100

display(hourly_analysis.sort_values(
    'fraud_rate', ascending=False
))
```

```python
plt.figure(figsize=(12, 5))

sns.barplot(
    x=hourly_analysis.index,
    y=hourly_analysis['fraud_rate']
)

plt.title('Fraud Rate by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Fraud Rate (%)')

plt.show()
```

```python
fraud_by_day = df.groupby('day_name')['is_fraud'].sum()

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

fraud_by_day = fraud_by_day.reindex(day_order)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=fraud_by_day.index,
    y=fraud_by_day.values
)

plt.title('Fraud Transactions by Day of Week')
plt.xlabel('Day')
plt.ylabel('Fraud Count')

plt.xticks(rotation=45)
plt.show()
```

```python
weekday_analysis = df.groupby('day_name').agg(
    total_transactions=('is_fraud', 'count'),
    fraud_transactions=('is_fraud', 'sum')
)

weekday_analysis['fraud_rate'] = (
    weekday_analysis['fraud_transactions'] /
    weekday_analysis['total_transactions']
) * 100

day_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

weekday_analysis = weekday_analysis.reindex(day_order)

display(weekday_analysis)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=weekday_analysis.index,
    y=weekday_analysis['fraud_rate']
)

plt.title('Fraud Rate by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Fraud Rate (%)')

plt.xticks(rotation=45)
plt.show()
```

```python
customer_analysis = df.groupby('cc_num').agg(
    total_transactions=('is_fraud', 'count'),
    total_amount=('amt', 'sum'),
    avg_transaction=('amt', 'mean'),
    fraud_transactions=('is_fraud', 'sum')
)

customer_analysis['fraud_rate'] = (
    customer_analysis['fraud_transactions'] /
    customer_analysis['total_transactions']
) * 100

display(
    customer_analysis.sort_values(
        'fraud_transactions',
        ascending=False
    ).head(10)
)
```

```python
overall_fraud_rate = df['is_fraud'].mean() * 100

customer_analysis['risk_category'] = np.select(
    [
        customer_analysis['fraud_rate'] > overall_fraud_rate * 2,
        customer_analysis['fraud_rate'] > overall_fraud_rate
    ],
    [
        'High Risk',
        'Medium Risk'
    ],
    default='Low Risk'
)

print(customer_analysis['risk_category'].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=customer_analysis,
    x='risk_category',
    order=['Low Risk', 'Medium Risk', 'High Risk']
)

plt.title('Customer Risk Category Distribution')
plt.xlabel('Risk Category')
plt.ylabel('Number of Cardholders')

plt.show()
```

```python
# Filter customers with at least 20 transactions

reliable_customers = customer_analysis[
    customer_analysis['total_transactions'] >= 20
].copy()

# Sort by fraud rate

display(
    reliable_customers.sort_values(
        'fraud_rate',
        ascending=False
    ).head(10)
)
```

```python
merchant_analysis = df.groupby('merchant').agg(
    total_transactions=('is_fraud', 'count'),
    total_amount=('amt', 'sum'),
    fraud_transactions=('is_fraud', 'sum')
)

merchant_analysis['fraud_rate'] = (
    merchant_analysis['fraud_transactions'] /
    merchant_analysis['total_transactions']
) * 100

display(
    merchant_analysis.sort_values(
        'fraud_transactions',
        ascending=False
    ).head(10)
)
```

```python
top_merchants = merchant_analysis.sort_values(
    'fraud_transactions',
    ascending=False
).head(10)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_merchants['fraud_transactions'].values,
    y=top_merchants.index
)

plt.title('Top 10 Merchants by Fraud Transactions')
plt.xlabel('Fraud Transactions')
plt.ylabel('Merchant')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['amt'], bins=50, kde=True)

plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Amount ($)')
plt.ylabel('Frequency')

plt.show()
```

```python
Q1 = df['amt'].quantile(0.25)
Q3 = df['amt'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['amt'] < lower_bound) |
    (df['amt'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Total Outliers:", len(outliers))
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='is_fraud',
    y='amt'
)

plt.title('Transaction Amount: Fraud vs Non-Fraud')
plt.xlabel('Fraud (0 = No, 1 = Yes)')
plt.ylabel('Transaction Amount ($)')

plt.show()
```

```python
amount_comparison = df.groupby('is_fraud')['amt'].agg(
    ['count', 'mean', 'median', 'max']
)

display(amount_comparison)
```
